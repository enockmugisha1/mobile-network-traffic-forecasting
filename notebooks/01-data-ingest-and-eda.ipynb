# Mobile network traffic forecasting - data ingest and EDA

This notebook loads the Milan telecom dataset, checks the file format, and begins the exploratory analysis required by the assignment.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data_dir = Path('..') / 'mobile-network'
print('Data directory exists:', data_dir.exists())
files = sorted(data_dir.glob('*.txt'))
print('Number of daily files:', len(files))
print('First few files:')
for f in files[:5]:
    print(f.name)
```

## Data format inspection

```python
# Count columns across a few files to understand structure
for p in files[:3]:
    df = pd.read_csv(p, sep=r'\s+', header=None, engine='python')
    print(p.name, 'columns=', df.shape[1], 'rows=', df.shape[0])

# Example of a single file's descriptive stats to verify sparsity
df = pd.read_csv(files[0], sep=r'\s+', header=None, engine='python')
print('Non-null count per column:')
print(df.notna().sum().to_string())
```
